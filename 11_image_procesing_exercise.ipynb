### Soal 1  
- Ambil sebuah gambar diri Anda (selfie) dengan kamera atau smartphone.  
- Lakukan cropping untuk menghasilkan:  
    - Cropping kotak persegi di wajah  
    - Cropping persegi panjang pada area latar belakang  
- Resize hasil crop ke ukuran 256×256 piksel.  
- Konversi gambar menjadi grayscal dan HSV, lalu tampilkan ketiganya berdampingan.  
- Tambahkan anotasi teks berisi nama Anda di atas kepala Anda pada gambar hasil crop.  
- Jelaskan efek cropping dan perubahan warna yang Anda lakukan.

### Soal 2  
- Gunakan gambar hasil crop pada Soal 1, konversikan warna gambar dalam ruang warna RGB.  
- Lakukan manipulasi channel warna:  
    - Naikkan intensitas channel merah sebanyak 50 poin (maks 255).  
    - Turunkan intensitas channel biru sebanyak 30 poin (minimum 0).  
- Gabungkan channel warna kembali dan simpan gambar hasil modifikasi.  
- Visualisasikan histogram distribusi warna untuk gambar asli dan hasil modifikasi.  
- Jelaskan dampak perubahan RGB pada warna gambar.

### Soal 3  
- Ambil gambar objek dengan background bertekstur (misal: kain bermotif, jerami, atau batu).  
- Terapkan filter edge detection (Canny) dan visualisasikan hasilnya.  
- Lakukan thresholding dengan nilai ambang tertentu supaya hanya objek utama tersisa.  
- Buat bounding box di sekitar objek hasil segmentasi.  
- Aplikasikan filter blur dan sharpening secara bergantian, lalu tampilkan hasilnya.  
- Jelaskan bagaimana setiap filter memengaruhi detail gambar.

### Soal 4  
- Rekam sebuah gambar diri Anda dengan ekspresi wajah netral.  
- Terapkan deteksi wajah dan landmark wajah menggunakan MediaPipe.  
- Buat overlay filter digital kreatif Anda sendiri (misal: topi, kumis, masker, helm, aksesoris unik lain) yang ditempatkan sesuai landmark wajah.  
- Gunakan alpha blending agar overlay terlihat natural di wajah.  
- Simpan hasilnya dan tampilkan gambar asli vs hasil filter.  
- Jelaskan metode perhitungan posisi overlay dan tantangan yang Anda temui selama implementasi.  

> **Catatan:** Sebagai contoh referensi, filter kacamata hitam dan hidung tomat merah pernah dibuat pada proyek sebelumnya  dan . Namun untuk tugas ini, Anda bebas berkreasi dengan desain overlay yang berbeda.

### Soal 5  
- Ambil gambar objek datar seperti karya seni tangan di kertas, tulisan di papan tulis, atau foto produk di meja dengan kondisi pencahayaan dan sudut tidak ideal.  
- Lakukan preprocessing untuk memperbaiki hasil tangkapan kamera agar tampak lebih rapi dan jelas:  
    - Konversi ke grayscale  
    - Koreksi perspektif (transformasi homografi agar objek terlihat sejajar dan terdistorsi minimal)  
    - Terapkan thresholding adaptif atau Otsu untuk memperjelas garis atau tekstur objek  
- Visualisasikan setiap tahap pemrosesan beserta hasil akhirnya.  
- Jelaskan fungsi masing-masing tahap dan bagaimana teknik ini membantu meningkatkan kualitas visual dan interpretasi citra.
